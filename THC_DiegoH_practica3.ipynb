```python
m=float(input('ingrese la masa del objeto: '))
v=float(input('ingrese la velocidad a la que se desplaza el objeto: '))
h=float(input('ingrese la altura a la que se encuentra el objeto: '))
g=9.81
Ec=(m*(v**2))/2
Ep=m*g*h
Em=Ec+Ep

print('Considerando un objeto de masa',m,'Kg, que se desplaza a una velocidad de',v,'m/s y que se ecunentra a una altura de',15,'m')
print('su energia cinetica es de',Ec,'J')
print('su energia potencial es de',Ep,'J')
print('su energia mecanica es de',Em,'J')
```

```python
import math
a=float(input('ingrese la medida de un lado a en centimetros: '))
b=float(input('ingrese la medida de un lado b en centimetros: '))
c=float(input('ingrese la medida de un lado c en centimetros: '))
c1=a+b
b1=a+c
a1=b+c
if c1>c and b1>b and a1>a:
  s=(a+b+c)/2
  A=math.sqrt((s*(s-a)*(s-b)*(s-c)))
  print('considerando un triángulo de lados: ',a,'cm,',b,'cm,',c,'cm')
  print('el semiperimetro del triángulo es de:',s,'cm')
  print('el área del triángulo es de:',A,'cm^2')
else:
  print('los lados ingresados no forman un triangulo')
```
