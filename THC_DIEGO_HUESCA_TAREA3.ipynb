```python
import math
print('ejercicio 2, perimetro de una elipse \n')
a=float(input('ingrese el valor del semieje mayor (a): '))
b=float(input('ingrese el valor del semieje menor (b): '))

#se analiza primero si no es un círculo porque la del círculo está de bolón
#pimpón
if a==b:
  print('La elipse es un círculo')
  P=math.pi*(2*a)
  print(f'El perímetro aproximado de la elipse es: {P: .3f}')

#si no es un círculo, empleamos la fórmula completa
else:
  c1=a+b
  c2=3*a+b
  c3=a+3*b
  P=math.pi*((3*c1)-math.sqrt(c2*c3))
  print(f'El perímetro aproximado de la elipse es: {P: .3f}')
```

```python
import math
print('ejercicio 9, volumen y área de la superficie de una pirámide regular recta \n')

#importante que la cantidad de lados sea un número entero ya que nimodo que
#tenga 10.4 lados
n=int(input('ingrese la cantidad de lados que tiene el polígono de la base de la pirámide: '))
l=float(input('ingrese la medida de un lado del polígono de la base de la pirámide: '))
h=float(input('ingrese la altura de la pirámide: '))

#analizamos que los datos sean correctos y referentes a una pirámide
if h>0 and n>=3 and l>0:

  #calculamos en base a las operaciones del ejercio
  #a=apotema A=área ap=apotema de la pirámide
  a=(((0.5)*l)*1/(math.tan(math.pi/n)))
  A=((0.5)*(n*l*a))
  ap=math.sqrt((h**2)+(a**2))

  V=(1/3)*A*h
  S=A+((1/2)*n*l*ap)

  print(f'el área de la superficie de la pirámide es de {S: .4f} cm^2')
  print(f'el volumen de la pirámide es de {V: .4f} cm^3')

else:
  print('ingresa valores válidos para una pirámide')
```

```python
import math
print('ejercicio 10, normalización de ángulos y determinación de cuadrantes\n')
a=float(input('ingrese el ángulo que desea normalizar: '))

#primero nos aseguramos de desplazar el ángulo a un ángulo entre 0 y 360
if a<0:

  #seguimos sumando 360 hasta que el ángulo se vuelva positivo
  while a<0:
    a=a+360
  ang=a

#si se pasa, aplicamos un protip que es dividir el ángulo entre 360 para
#remover todas las "vueltas completas al círculo" y nos quedamos únicamente con
#el residuo
elif a>360:
  ang=a%360

#si ya está entre 0 y 360 pues ya estuvo
else:
  ang=a

#evaluamos si es múltiplo de 90 por lo que estaría en alguno de los ejes
if ang%90==0:
  print('el ángulo reducido cae exactamente en un eje')

  #averiguamos en cuál eje
  if ang==0 or ang==360:
    print('el ángulo cae en el eje x positivo')
  elif ang==90:
    print('el ángulo cae en el eje y positivo')
  elif ang==180:
    print('el ángulo cae en el eje x negativo')
  #en los pasos anteriores nos aseguramos de que ang estuviera en el rango
  #entre 0 y 360, por lo tanto los 270 grados son el else basandonos en la
  #condición de que sea un numero divisible entre 90
  else:
    print('el ángulo cae en el eje y negativo')

#calculamos el cuadrante en el que se encuentra el ángulo llendo del "menor" al
#"mayor" ya que existen angulos que son menores que 270 y que 90, por lo tanto,
#si hicieramos el cálculo iniciando con esa evaluación, lo más probable, es que
#la mayoría de los ángulos estarían en el tercer cuadrante
elif ang<90:
  print('el ángulo cae en el primer cuadrante')
elif ang<180:
  print('el ángulo cae en el segundo cuadrante')
elif ang<270:
  print('el ángulo cae en el tercer cuadrante')
else:
  print('el ángulo cae en el cuarto cuadrante')

print('el ángulo reducido es de',ang,'grados')
```
