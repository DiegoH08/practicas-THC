```python
z1 = 8 + 5j
z2 = 1 - 1j

suma = z1 + z2

resta = z1 - z2

conjugado_z1 = z1.conjugate()

print("Suma (z1 + z2):", suma)
print("Resta (z1 - z2):", resta)
print("Conjugado de z1:", conjugado_z1)
```

```python
x = float(input('ingresa tu primer numero: '))
y = float(input('ingresta tu segundo numero: '))

suma= x + y
resta= x - y
producto= x*y
division= x/y
potencia= x**y
divZ= x//y
residuo= x%y

print('la suma de esos numeros es', suma)
print('la resta de esos numeros es', resta)
print('el producto de esos numeros es', producto)
print('la division de esos numeros es', division)
print('la potencia de esos numeros es', potencia)
print('la division entera de esos numeros es', divZ)
print('el residuo de la division de esos numeros es', residuo)
```
